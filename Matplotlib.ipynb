```python
import matplotlib.pyplot as plt
```

## Line Plot

```python
months=[1,2,3,4,5]
sales=[100,120,150,170,200]
```

```python
plt.plot(months, sales)
```

```python
plt.plot(sales,months)
```

```python
plt.plot(months,sales)

plt.title("Sales Growth")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
```

```python
plt.xlabel("Months")
plt.ylabel("Sales")
plt.plot(months,sales,marker="*")
plt.show()
```

```python

plt.plot(months,sales,linestyle="--")
plt.show()
```

```python
plt.plot(months,sales,linestyle="-.")
plt.show()
```

##Multiple Lines

```python
profit=[30,45,60,65,70]
plt.plot(months,sales,label="Sales",color="Pink")
plt.plot(months,profit,label="Profit",color="Blue")

plt.legend()
plt.show()
```
