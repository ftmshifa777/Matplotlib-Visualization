##Bar Chart

```python
subjects=["Python","SQL","ML","DL"]
marks=[50,90,75,67]

plt.bar(subjects,marks,color='mediumseagreen')

plt.title("Subject Marks")
plt.xlabel("Subjects")
plt.ylabel("Marks")

plt.show()

```

##Scatter Plot

```python
hrs=[1,4,7,8,9,4,6,2,5,10]
marks=[80,44,66,86,75,22,45,67,89,91]

plt.scatter(hrs,marks,marker='s',color='violet',linewidth=2,edgecolor='blue')
plt.title("Study Hours VS Marks")
plt.xlabel("Hours Studied")
plt.ylabel("Marks")
plt.show()
```

##Pie Chart

A pie chart represents parts of a single whole.

Pie charts become difficult to read when there are many categories or when exact comparison matters. A bar chart is often clearer.

```python
sub=["Python","SQL","ML","DL"]
hrs=[30,10,20,40]


plt.pie(hrs,labels=sub)
plt.show()

```

```python
sub=["Python","SQL","ML","DL"]
hrs=[30,10,20,40]


plt.pie(hrs,labels=sub,shadow=True,hatch='\/',startangle=30,wedgeprops=dict(width=0.5))
plt.show()
```

##Histogram

A Histogram shows the distibution of numerical data. It groups values into interval called bins.

```python
marks=[87,23,45,66,43,77,44,66,33,40,89,10,26]

plt.hist(marks,bins=5,color='orange',edgecolor='blue',linewidth='4')
plt.xlabel("Marks")
plt.ylabel("Marks Distribution")

plt.show()
```

##Box Plots

A box plot summarises the didtribution of numerical data and helps identify potential outliers.

```python
marks=[45,69,79,32,89,35,255,84,100,89,101]

plt.boxplot(marks)
plt.title("Marks Box Plot")
plt.show()
```

##Grid

It adds horizontal and vertical grid lines into the graph, making it easier to read values.

```python
from matplotlib.lines import lineStyles
marks=[45,50,60,70,80,90]
plt.plot(marks)

plt.xlabel("Students")
plt.ylabel("Marks")
plt.title("Student Marks")

plt.grid(color='violet',linestyle='--',linewidth=1)
plt.show()

```

##Figure Size

```python
marks=[45,50,60,70,80,90]

plt.figure(figsize=(8,4))
plt.plot(marks)

plt.xlabel("Students")
plt.ylabel("Marks")
plt.title("Student Marks")

plt.show()

```

```python
age=[12,20,15,17,19,11,18,17,19,13]
plt.figure(figsize=(10,8))

plt.plot(age,color='green',linestyle='--')
plt.xlabel("Children")
plt.ylabel("Age")
plt.title("Ages of Children")
plt.show()
```

##Axis Limits

Axis limits control how much of the graph is visible.

There are two main areas:

X-axis -> horizontal

Y-axis -> vertical

```python
plt.plot(marks)
plt.xlim(0,6)
plt.ylim(0,100)

plt.xlabel("Students")
plt.ylabel("Marks")
plt.title("Student Marks")

plt.show()
```

```python
plt.plot(age,color='orange',linestyle='-.')
plt.ylim(10,25)
plt.xlim(0,15)

plt.xlabel("Teens")
plt.ylabel("Ages")

plt.show()

```

##Rotating Labels

Rotating Labels means changing the angle of the text shown on the X-axis or Y-axis.

```python
subjects=["Python","Machine Learning","Deep Learning","NLP"]
marks=[80,75,90,85]

plt.bar(subjects, marks,color='purple')

plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.title("Student Marks")

plt.xticks(rotation=45)

plt.show()
```

```python
plt.bar(subjects,marks,color='skyblue')
plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.xticks(rotation=90)
plt.show()
```

```python
plt.bar(subjects,marks,color='maroon')
plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.xticks(rotation=0)
plt.show()
```

plt.xticks(rotation=0) #Normal

plt.xticks(rotation=45) #45 degrees

plt.xticks(rotation=90) #Vertical





Rotation makes axis labels easier to read when they are long
